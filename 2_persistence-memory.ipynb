2__Fault_tolerance

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langgraph.checkpoint.memory import InMemorySaver
import time
```

```python
class crashstate(TypedDict):
    input:str
    step1:str
    step2:str
    step3:str
    
```

```python
def step_1(state:crashstate)->crashstate:
    print("step_1 executed")
    return {
        'step1':'done', 'input':state['input']
    }
```

```python
def step_2(state:crashstate)->crashstate:
    print("step_2 hanging....now manually interrupt from thr notebook tool bar (stop button) ")
    time.sleep(30)
    return {
        'step2':'done'
    }

def step_3(state:crashstate)->crashstate:
    print("step_3 executed")
    return {
        'step3':'done'
    }
```

```python
graph=StateGraph(crashstate)

graph.add_node('step_1', step_1)
graph.add_node('step_2', step_2)
graph.add_node('step_3', step_3)

graph.add_edge(START, 'step_1')
graph.add_edge('step_1', 'step_2')
graph.add_edge('step_2', 'step_3')
graph.add_edge('step_3', END)


checkpointer=InMemorySaver()
workflow=graph.compile(checkpointer=checkpointer)

```

```python
try:
    print('running grpah.... manually interrupy during step_2')
    config1= {'configurable':{'thread_id': '1', 'checkpoint_id': '1f1a1e7d-9aeb-600b-8001-2335b4f99418'}}
    workflow.invoke({None }, config=config1,)
except KeyboardInterrupt:
    print('kernal manually interrupt (crash simulated)')
```

```python
workflow.get_state(config1)
```

```python
list(workflow.get_state_history(config1))
```

```python
#re_run to show fault tolerance resume
#and wahi step se dobara resume karenge jaha pe crash hoagay (ste:2)
final_state=workflow.invoke(None,config=config1)
print("final:",final_state)
```
