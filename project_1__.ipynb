```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()

model=ChatGroq(model="openai/gpt-oss-120b")
```

```python
class jobstate(TypedDict):
    jd:str
    my_skill:str
    requirements:list[str]
    match_scre:int
    output:str
```

```python
def extract_req(state:jobstate)->jobstate:
    job=state['jd']

    prompt=f"""extract the requirements from the given job description - {job} """
    response= model.invoke(prompt)

    return {
        'requirements':[response]
    }
```

```python
def  match_skills(state:jobstate)->jobstate:
    require=state['requirements']
    my_skill=state['my_skill']

    prompt=f"""match the requirement {require} and my skills {my_skill} and generate the match score for it"""
    response=model.invoke( prompt)

    return {
        'match_scre':response
    }
```

```python
def gen_letter(state:jobstate)->jobstate:
    jd= state['jd']
    my_skill=state['my_skill']

    prompt=f"""write a professional cover letter for this job:
    job description: {jd}
    my skills: {my_skill}

    rules:
    - keep it under 500 words
    - highlight matching skill
    - professional tone"""

    response= model.invoke(prompt)

    return {
        'output' : response
    }
```

```python
def suggest_imp(state:jobstate)->jobstate:
    jd= state['jd']
    my_skill=state['my_skill']

    prompt=f"""suggest improvement for the job:
    job description: {jd}
    my skills: {my_skill}"""

    response=model.invoke(prompt)

    return {
        'output':response
    }
```

```python
def decision(state:jobstate)->jobstate:
    if state['match_scre']>=60:
        return 'gen_cover_letter'
    else: 
        return 'suggest improvement'
```

```python
graph=StateGraph(TypedDict)

#nodes
graph.add_node('extract_req', extract_req)
graph.add_node('match_skills', match_skills)
graph.add_node('gen_letter', gen_letter )
graph.add_node('suggest_imp', suggest_imp )

#edges
graph.add_edge(START, )
```
